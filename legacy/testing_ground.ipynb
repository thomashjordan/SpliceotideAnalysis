```python
# =============================================================================
# CELL 1: IMPORTS & CONFIGURATION
# =============================================================================
import sys
import os
import pickle
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Make sure the project root is in the Python path to find local modules
project_root = Path(os.getcwd()).parent
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from pipeline_functions import run_new_analysis_for_group
from classes_spectrometry import *
from peptide_classes import *
from analysis_containers import ReplicateGroupResults, GroupedExperiments

INPUT_EXCEL = Path(r'D:Spliceotide_Data_Analysis\input_excel_files\flask_expressions.xlsx')
OUTPUT_PATH_DIR =Path(r'D:Spliceotide_Data_Analysis\output_data_flasks')
REPORTS_OUTPUT_DIR = Path(r'D:Spliceotide_Data_Analysis\output_reports_flasks')

# Set pipeline behavior flags
FORCE_RELOAD = False     # True to re-parse .mzML files, ignoring spectral cache.
FORCE_REANALYZE = True   # True to re-run peak picking, ignoring master data cache.
MANUAL_REVIEW = False    # True to open interactive plots for manual peak validation.
```

```python
# =============================================================================
# CELL 2: FIND GROUP AND RUN ANALYSIS
# =============================================================================
print("--- Step 1: Finding Experimental Group ---")

if not INPUT_EXCEL.exists():
    raise FileNotFoundError(f"Input Excel file not found at: {INPUT_EXCEL}")
df_samples = pd.read_excel(INPUT_EXCEL)

# Get the unique identifiers for the entire experimental group
group_keys = ['experiment_id', 'precursor_plasmid', 'maturase_name', 'maturase_plasmid', 'expression_temperature', 'digestion']

# Isolate the full DataFrame for this experimental group (all replicates)
selected_df = df_samples[df_samples["maturase_plasmid"]=="TJ-C-689"]
selected_group_df = selected_df[(selected_df["precursor_plasmid"]=="TJ-C-614")
                                 & (selected_df["digestion"]=="trypsin")]

grouped = selected_group_df.groupby(group_keys)
group_info_tuple = tuple(grouped.groups.keys())[0]  # Get the first (and only) group key tuple
cohort_results = run_new_analysis_for_group(
    group_info=group_info_tuple,
    group_df=selected_group_df,
    report_output_dir=REPORTS_OUTPUT_DIR,
    force_reload=FORCE_RELOAD,
    manual_review=MANUAL_REVIEW,
    force_reanalyze=FORCE_REANALYZE
    )

print("\n--- Analysis run complete. ---")
```
